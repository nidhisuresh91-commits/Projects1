## Identify Missing Values

```python
#Detect missing values in the dataset.
student_data.isnull().sum()
```

**Columns containing missing values :** Age, Hours_Studied, Attendance, Parental_Education, Final_Score

**Columns which need mean filling :** Age, Hours_Studied, Attendance, Final_Score

**Columns which need mode filling :** Parental_Education

## Handle Missing Values using fillna()

```python
#For numerical columns (Age, Hours_Studied, Attendance, Final_Score):
student_data['Age']= student_data['Age'].fillna(student_data['Age'].mean())
student_data['Hours_Studied']= student_data['Hours_Studied'].fillna(student_data['Hours_Studied'].mean())
student_data['Attendance']= student_data['Attendance'].fillna(student_data['Attendance'].mean())
student_data['Final_Score']= student_data['Final_Score'].fillna(student_data['Final_Score'].mean())
```

```python
# For categorical columns (Gender, Parental_Education):
student_data['Parental_Education']= student_data['Parental_Education'].fillna(student_data['Parental_Education'].mode()[0])
```

```python
#verify
student_data.isnull().sum()
```

## Encode Categorical Variables

```python
#Encode Gender using Label Encoding
le = LabelEncoder()
student_data['Gender'] = le.fit_transform(student_data['Gender'])
le.classes_
```

```python
student_data.head()
```

```python
#Apply One-Hot Encoding to Parental_Education using Pandas.
student_data = pd.get_dummies(student_data, columns=['Parental_Education'])
student_data.head()
```
